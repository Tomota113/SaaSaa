```python
#Normaliser les images pourqu'elles aient des valeurs entre 0 et 1
train_images = train_images.reshape((-1,28*28))
test_images= test_images.reshape((-1,28*28))
```

```python
from keras.utils import to_categorical
#convertir les etiquettes en categories one_hot
train_labels = to_categorical(train_labels)
test_labels= to_categorical(test_labels)
```

```python
from keras.models import Sequential
from keras.layers import Dense,Flatten

#Construire le modele MLP
model = Sequential()
#couche d'entree
model.add(Flatten(input_shape=(28*28,)))
# Couches cachees
model.add(Dense(128,activation='relu'))
model.add(Dense(64,activation='relu'))
model.add(Dense(32,activation='relu'))
#Couche de sortie
model.add(Dense(10,activation='softmax'))

#Configurer le modele
model.compile(optimizer='Adam', loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
import time

start_time=time.time()

#Entrainer le modele
model.fit(train_images,train_labels,epochs=100)

#Calculer et afficher le temps ecoule
elapsed_time=time.time() - start_time
print(f"Temps d'entrainement : {elapsed_time:.2f} secondes")

#evaluer le modele
test_loss, test_acc = model.evaluate(test_images, test_labels)
print("Test accuracyt: ",test_acc)
```
